# Data deduplication with Python

As you’ve learned, the data cleaning and validating practices include several different steps, including handling missing data, outliers, and label encoding; checking for misspellings; and, handling duplicates. As a data professional, it will be your task to know how best to handle data values in those categories. In this reading, you'll learn more about handling duplicates. You will also learn to identify and decide whether deduplication is the right strategy for a dataset. In addition, you will learn some common Python functions for handling duplicates.

## Identifying duplicates 

Before we make any decisions on whether to remove duplicate values or not, we should first determine if duplicate values are present in our dataset. 

A simple way to identify duplicates is to use the **duplicated()** function from Pandas. **duplicated()** is a method of the **DataFrame** class.

This function returns a series of “true/false” outputs, with “true” indicating the data value is a duplicate, and “false” indicating it is a unique value. 

Here’s an example of a five-row dataframe: 

````python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'brand': ['Wowyow', 'Wowyow', 'Splaysh', 'Splaysh', 'Pipplee'],
    'style': ['cistern', 'cistern', 'jug', 'stock', 'stock'],
    'rating': [4.0, 4.0, 5.5, 3.3, 3.0]
})

df
````

Using the **duplicated()** function, the result is that one has been marked “True,” indicating it is a duplicate.

````python
df.duplicated()
````

Identifying duplicates for an entire dataframe will be different than a single column or index. Be sure when you use the **duplicated()** **function for an entire dataframe**. The **duplicated()** function will only return *entire rows* that have exactly matching values, not just individual matching values found within a column.

If you wish to identify duplicates for only one column or a series of columns within a dataframe, you will need to include that in the “subset” portion of the argument field of the **duplicated()** function. Going further, if you’d like to specify which of the duplicates to keep as the “original” as opposed to the duplicate, you can specify that in the **keep** portion of the argument field.

Below is an example of identifying duplicates in only one column (subset) of values and labeling the last duplicates as “false,” so that they are “kept”: 

````python
print(df)
print()
print(df.duplicated(subset=['style'], keep='last'))
````

## Decision time: To drop or not to drop?

As you’ve learned, every dataset is unique and you cannot treat every dataset the same. When you are making the decision on whether to eliminate duplicate values or not, think deeply about the **dataset itself** and about **the objective you wish to achieve**. What impact will dropping duplicates have on your dataset and your objective? 

**1. Deciding to drop**

You should **drop or eliminate** duplicate values if duplicate values are clearly mistakes or will misrepresent the remaining unique values in the dataset.  

For example, you can be reasonably sure that a data professional will (in most cases) eliminate duplicate values of a dataset containing house addresses and house prices. Counting the same house twice will (in most cases) misrepresent any conclusions drawn from the dataset as a whole, such as average house price, total house price, or even total number of houses. In a case like this, a data professional would almost certainly eliminate the duplicate data so as to fairly represent the remaining data during analysis and visualization. 

**2. Deciding to NOT drop**

You should **keep** duplicated data in your dataset if the duplicate values are clearly **not** mistakes and should be taken into account when representing the dataset as a whole. 

For example, a dataset marking the number of throws and distances of an Olympic shot-put athlete in training will likely include several duplicate distances; just by nature of number of attempts and the limits a person can have a weighted ball, there will be duplicate values—particularly if the distance measurements are labeled to only 1 or 2 decimal places. In a case like this, a data professional would almost certainly keep all of the data to fairly represent it as a whole during analysis and visualization.   

## Don’t be duped — How to do deduplication

Before we get back into Python and learn how to eliminate duplicates, let’s first define the term “deduplication”: 

-   **Deduplication:**  The elimination or removal of matching data values in a dataset.

There are a number of different libraries, functions, and methods in Python you could use to remove matching data values. 

One of the more common functions to use is in Pandas: **drop_duplicates()**

**drop_duplicates()** is another **DataFrame** method. It's used to create a new dataframe with all of the duplicate rows removed. 

For example, use a dataframe from earlier in this reading: 

````python
df.drop_duplicates()
````

You’ll notice in the resulting output that the duplicate row of data was removed, leaving the remaining unique values intact. 

**Note:** Keep in mind that the **drop_duplicates()** function as written above will only drop duplicates of exact matches of **entire rows of data**. If you wish to drop duplicates within a single column, you will need to specify which columns to check for duplicates using the **subset** keyword argument.

This example drops all rows that have duplicate values in the **style** column (except for the first occurrence):

````python
print(df)
df_dedup = df.drop_duplicates(subset='style')

print()
print(df_dedup)
````

And this example drops all rows (except the first occurrence) that have *duplicate values in* *both* the **style** and **rating** columns:

````python
print(df)
df = df.drop_duplicates(subset=['style', 'rating'])

print()
print(df)
````

In data analysis, "missing data" isn't a single concept. Depending on whether you are working in SQL, pure Python, NumPy, or Pandas, missing values are represented differently and behave under different rules.

## Detailed Breakdown

### `NULL` (SQL & Python `None`)

-   **What it is:** The universal database and computer science standard for an unknown or missing value.
-   **In Python:** Expressed as `None` (a singleton object of type `NoneType`).
-   **In SQL:** Expressed as `NULL`.
-   **Key Behavior:** `NULL` is not equal to anything, not even itself. In SQL, `NULL = NULL` evaluates to `UNKNOWN` (hence requiring `IS NULL`). In Pandas, `None` in numeric columns automatically gets converted to `NaN`.

````python
# Python behavior
print(None == None)  # True (Python singleton identity)

# SQL logic simulation
# NULL = NULL -> FALSE / UNKNOWN
````

The logic behind the SQL logic where NULL is not equal to NULL is because null represents *unknown* and we cannot asset that one unknown value equals another unknown value.

### `NaN` (Not a Number)

-   **What it is:** A numerical missing value standard defined by IEEE 754 floating-point arithmetic.

-   **In Python / Pandas:** Derived from NumPy (`import numpy as np; np.nan`).

-   **Key Behavior:**

    1.  It is strictly a **floating-point** value.

    2.  **Type coercion side-effect:** If a standard integer column contains even a single `NaN`, Pandas converts the entire column to `float64`.

    3.  `NaN` is **never equal to itself**: `np.nan == np.nan` evaluates to `False`!

````python
#| md-indent: '        '
# Inequality trap
print(np.nan == np.nan)  # False!

# How to check for NaN correctly:
print(np.isnan(np.nan))  # True
print(pd.isna(np.nan))   # True
````

### `NA` (Not Available)

-   **What it is:** A missing value indicator introduced in modern Pandas (`pd.NA`, starting in version 1.0) modeled after R's `NA`.

-   **Why it was created:** To fix the `NaN` float conversion problem. `pd.NA` works across nullable data types (like `Int64`, `boolean`, and `string`) without forcing integers to floats or booleans to object types.

-   **Key Behavior:** Implements **three-valued logic** (propagates through logical operations).

````python
#| md-indent: '    '
# Enables integer columns to retain missing values without becoming floats
s = pd.Series([1, 2, None], dtype="Int64")  # Capital 'I' Nullable Integer type

print(s)
# 0       1
# 1       2
# 2    <NA>
# dtype: Int64
````

### `Blank` (Empty String `""` or Whitespace `" "`)

-   **What it is:** A text string containing zero characters (`""`) or only space characters (`"   "`).

-   **Crucial Distinction:** **A blank string is NOT missing data.** To a computer, a blank string is a perfectly valid piece of string data that happens to be empty.

-   **The Danger:** Pandas `df.isna()` or `df.isnull()` will return `False` for empty strings. They will quietly bypass missing-data filters and pollute your analysis.

````python
#| md-indent: '    '
df_text = pd.DataFrame({'city': ['Madrid', '', '   ', None]})

# Pandas does NOT consider empty strings or spaces as missing!
print(df_text['city'].isna())
# 0    False  (Valid)
# 1    False  (Empty string - NOT detected as missing!)
# 2    False  (Spaces - NOT detected as missing!)
# 3     True  (None - Correctly detected as missing)
````

## Practical Example: How Pandas Handles All Four

````python
# Creating a DataFrame with all 4 missing/empty types
demo_df = pd.DataFrame({
    'sql_none': [1, None, 3],
    'numpy_nan': [1.0, np.nan, 3.0],
    'pandas_na': pd.Series([1, pd.NA, 3], dtype="Int64"),
    'blank_string': ['apple', '', 'banana'],
    'whitespace': ['apple', '   ', 'banana']
})

demo_df
````

````python
# Notice how blank strings and whitespaces are NOT detected as missing!
demo_df.isna()
````

To properly clean text datasets containing empty strings or whitespace, you must convert them to `np.nan` before running `.isna()` or `.dropna()`:

````python
# Step 1: Strip whitespace and replace empty strings with np.nan
cleaned_df = demo_df.replace(r'^\s*$', np.nan, regex=True)

# Step 2: Now .isna() correctly detects all empty values
cleaned_df.isna()
````

### Why converting missing strings into nan and not na or null?

While `pd.NA` or `None` sound conceptually more accurate for text ("not applicable" or "null"), we convert to **`np.nan`** in traditional Pandas workflows for three practical reasons: **ecosystem compatibility**, **underlying column types**, and **performance**.

#### Third-Party Ecosystem Compatibility (The Biggest Reason)

Pandas doesn't exist in a vacuum—it is built on top of **NumPy** and feeds into libraries like **Scikit-Learn, SciPy, Matplotlib, and Seaborn**.

-   **NumPy and C-libraries** only recognize `np.nan` as missing data.

-   If you pass a DataFrame containing `None` or `pd.NA` into a Machine Learning model (like `scikit-learn`'s `SimpleImputer` or `XGBoost`), or a C-accelerated plotting library, it will often throw a `TypeError` because those C-extensions expect the standard IEEE 754 `np.nan` floating-point indicator.

#### Default `object` Dtypes in Pandas

By default, when you load text into Pandas, the column type is **`object`** (a column holding pointers to raw Python objects), not Pandas' newer experimental `string` type.

-   If you insert **`None`** into an `object` column, it stays as the literal Python `None` object.

-   If you insert **`pd.NA`** into a standard `object` column, Pandas treats it as a generic, custom Python object.

-   **`np.nan`**, however, is natively handled across all standard Pandas indexing routines, vectorized string operations (`.str`), and drop methods regardless of whether the column is numeric or `object`.

### Vectorization & Performance (Why `None` hurts)

When Pandas sees `np.nan`, it can delegate calculations to fast C-level code under the hood.

When a column contains Python `None`, Pandas is often forced to drop back into pure Python looping (iterating item-by-item through Python memory), which can make string cleaning on large datasets 10x to 100x slower.

#### Comparison: What Happens Under the Hood?

````python
# Standard object column (default text behavior)
s_nan  = pd.Series(['apple', np.nan])
s_none = pd.Series(['apple', None])
s_na   = pd.Series(['apple', pd.NA])

# 1. Scikit-Learn / Matrix operations
# s_nan  -> Works out of the box with ML imputers
# s_none -> Fails or forces slow object-evaluation
# s_na   -> Fails with TypeError in non-Pandas libraries

# 2. Vectorized NumPy routines
# np.array(s_nan)  -> Fast NumPy array
# np.array(s_none) -> Slow Python Object array
````

#### The Modern Exception: `pd.NA` & PyArrow

The rules are shifting slightly with **Pandas 2.0+**:

If you explicitly use Pandas' modern **nullable types** or the **PyArrow backend** (e.g., `dtype="string[pyarrow]"` or `dtype="Int64"`), **`pd.NA` IS the correct choice** because PyArrow natively understands null bitmaps without converting columns to `object` or `float`.

````python
import pyarrow as pa
# Modern Pandas 2.0+ approach:
s = pd.Series(['apple', ''], dtype="string[pyarrow]")

# In PyArrow string columns, empty strings convert to pd.NA seamlessly
s = s.replace('', pd.NA)
````

However, non-string types introduce a few **data-type specific rules** and behaviors you need to know.

### Behavior Across Non-String Data Types

#### Numeric Data (`float` vs `int`)

-   **Floats (`float64`)**: `np.nan` is a native floating-point value. It works seamlessly out of the box.

-   **Standard Integers (`int64`)**: Standard integer columns **cannot** store `np.nan`. If you introduce `np.nan` into an integer column, Pandas will silently upcast the entire column to `float64`.

````python
#| md-indent: '    '
s = pd.Series([1, 2, 3])  
s.dtype # dtype: int64
s[1] = np.nan                   # Type changed! dtype: float64 -> [1.0, NaN, 3.0]
s.dtype
````

**The Exception (`Int64`):** If you use Pandas' capital-I nullable integer type (`dtype="Int64"`), missing values are represented as **`pd.NA`**, preserving the integer type.

#### Datetime Data (`datetime64`)

Pandas does **not** use `np.nan` for missing dates/timestamps. It has a dedicated singleton called **`pd.NaT`** ("Not a Time").

-   If you assign `np.nan` or `None` to a datetime column, Pandas will **automatically convert it to `pd.NaT`** behind the scenes.

````python
dates = pd.to_datetime(['2026-01-01', None, np.nan])

print(dates)
# DatetimeIndex(['2026-01-01', 'NaT', 'NaT'], dtype='datetime64[ns]')
````

#### Boolean Data (`bool`)

Standard NumPy booleans can only ever be `True` or `False`.

-   If you put `np.nan` into a standard boolean column, Pandas degrades the entire column to **`object`** dtype (storing raw Python pointers, which ruins performance).

-   If you need missing values in booleans, use Pandas' nullable boolean type (`dtype="boolean"`), which uses **`pd.NA`**.

​A quick check of the non-null count column helps us confirm which ​columns have missing data, zip code, city, state, and state code. ​In fact, if we subtract the total number of strikes x and ​the number of strikes y the result is the number ​we identified earlier, 393,830. ​

Let's take a look at the top portion of the new data frame we created,  using the head function.

The output tells us exactly what we expected to find in the data ​frame we created. ​In the top five lines of data, the columns zip code, ​city, state, state code and ​number of strikes y have the NaN values in the cells. ​Now that we have pinpointed exactly what values are missing in our data frame. ​The last thing we'll do in this notebook is learn how these missing values ​impact our data. 

​The best way to do that is to create a data visualization. ​A plotted map will help us see where the majority of the missing values ​are located geographically. ​To design the map, let's start by creating another data frame. ​This one we'll call top_missing. ​With this data frame we are gathering only the data coms ​we will need in order to plot a geographic visualization, ​latitude, longitude, and number of strikes x. ​As you'll recall from earlier in this lesson, ​number of strikes x includes all of the 717,530 ​data rows while number of strikes y are a segment of those ​717,530 with missing data in the zip code, ​state, and state code columns. ​You'll find that it's helpful to group the columns by latitude first then longitude. 

​This will make it easier for most data processors to plot. ​Lastly, we'll sort the lightning strikes x column by some of values. ​In the ascending field will input, False, because we want ​the largest sums of lightning strikes at the top of the data frame. ​Finally, let's take a look at what we built using the head function again. 

````python
# Create new df of just latitude, longitude, and number of strikes and group by latitude and longitude
top_missing = df_null_geo[['latitude','longitude','number_of_strikes_x']
            ].groupby(['latitude','longitude']
                      ).sum().sort_values('number_of_strikes_x',ascending=False).reset_index()
top_missing.head(10)
````

​After running the cell, the first 10 rows show the most strikes ​are falling on and around the same latitudes and longitudes. ​Let me give you a helpful hint before we plot a visualization using this ​data frame. ​You should import the express package from plotly first. 

​Express is a helpful package that speeds up coding by doing a lot of ​the back-end work for you. ​If we don't use express for this particular data set which has hundreds ​of thousands of points to plot your run cell times could be long or ​the code could even break. ​Now, it's time to create the map. ​Our format for the graph will be from plotly express called scatter geo. ​

````python
fig = px.scatter_geo(top_missing[top_missing.number_of_strikes_x>=300],
                    lat = 'latitude',
                    lon = 'longitude',
                    size = "number_of_strikes_x");
fig.update_layout(
    title_text ="Missing data"
);
fig.show()
````

As indicated by its name, this graph type is used for ​scatter plots on a geographic map. ​In the argument field will input the data frame we just created, top missing. ​For the parameters in the argument field, ​we want to filter the number of values to be only those latitudes and ​longitudes with lightning strikes of more than 300. 

​Naturally, we plug latitude and longitude into their appropriate parameter spots. ​Then for the size parameter, we plug in the final column, number of strikes x. ​Lastly, let's use the update layout function within ​the figure to give the plot title, Missing data. ​Now, we create it. ​It's a nice geographic visualization but we really don't need the global scale. ​Let's scale it down to only the geographic area that we are interested in, ​the United States. ​So, let's copy the same code we just wrote, ​this time though we add the parameter geo_scope= 'usa' ,. 

````python
fig = px.scatter_geo(top_missing[top_missing.number_of_strikes_x>=300],
                    lat = 'latitude',
                    lon = 'longitude',
                    size = "number_of_strikes_x");
fig.update_layout(
    title_text ="Missing data",
    geo_scope='usa'
);
fig.show()
````

This will limit the geographic scope to just the United States of America. ​Then we can plot it again and check the output. ​The resulting map shows a majority of these missing values cropping ​up along the borders or in spots over bodies of water like lakes, ​the ocean, or the Gulf of Mexico. ​Given that the missing data were in the state abbreviations and ​zip code columns, it does make sense why those data points were left blank. ​There are no zip codes for bodies of water. ​You'll also notice some other locations with missing data that are not ​over bodies of water. ​These types of missing data, latitude and longitude on land ​are the kind of missing data you would want to reach out to the NOA A about. 

​What's interesting about cleaning the lightning strike data and looking for ​missing data values is that we learn something new in the process. ​We found one of the stories hidden in the data. ​The data values were missing because the lightning strikes were over water. ​It goes to show you that data cleaning may sound tedious but ​you'll never know what you learn.

## Missing data

The following pandas functions and methods are helpful when dealing with missing data.

### df.info()

Description: A DataFrame method that returns a concise summary of the dataframe, including a ‘non-null count,’ which helps you know the number of missing values

````python
data = {
    'planet': [
        'Mercury',
        'Venus',
        'Earth',
        'Mars',
        'Jupiter',
        'Saturn',
        'Uranus',
        'Neptune',
    ],
    'radius_km': [2440, 6052, 6371, 3390, 69911, 58232, 25362, 24622],
    'moons': [0, 0, 1, 2, 80, 83, 27, 14],
}

df = pd.DataFrame(data)

print(df)
print()
df.info()
````

### pd.isna() / pd.isnull()

Description: pd.isna() is a pandas function that returns a same-sized Boolean array indicating whether each value is null (you can also use pd.isnull() as an alias). Note that this function also exists as a DataFrame method.

````python
data = {
    'planet': [
        'Mercury',
        'Venus',
        'Earth',
        'Mars',
        'Jupiter',
        'Saturn',
        'Uranus',
        'Neptune',
    ],
    'radius_km': [2440, np.nan, 6371, 3390, 69911, np.nan, 25362, 24622],
    'moons': [0, 0, None, 2, 80, None, 27, 14],
}
pd.isnull(df)
````

### pd.notna() / pd.notnull()

### df.fillna()
